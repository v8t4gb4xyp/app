```python
import json

import requests

r = requests.post(
    "http://localhost:8000/invocations",
    headers={"Content-Type": "application/json"},
    data=json.dumps(
        {
            "input": [
                {
                    "role": "user",
                    "content": "How many tshirts do we have in stock in all of DC?",
                }
            ],
            "custom_inputs": {"model_id": "foobarz", "table": "order_items"},
        }
    ),
)

print(r.status_code, r.text)
```
